```python
#total count of employees who left (Attrition == 'Yes') vs. total employees.
employees_left = df[df['Attrition'] == 'Yes']['Attrition'].count()
print(f"employees left: {employees_left}")
employees_stayed = df[df['Attrition'] == 'No']['Attrition'].count()
print(f"employees stayed : {employees_stayed}")
```

```python
Total_employees = len(df)
print(f"total employees: {Total_employees}")
```

```python
#exact percentage of employees who left.
attrition_percentage = (employees_left/Total_employees) * 100
print(f"attrition_percentage: {attrition_percentage}")
```

```python
Jobrole_summary = pd.crosstab(df['JobRole'], df['Attrition'], margins=True)

Jobrole_summary['Attrition%'] = (Jobrole_summary['Yes'] / Jobrole_summary['All']) * 100
Jobrole_summary = Jobrole_summary.sort_values(by='Attrition%', ascending=False)
print(Jobrole_summary)
```

```python
dept_summary = pd.crosstab(df['Department'], df['Attrition'], margins=True)
print(dept_summary)
```

```python
overtime_summary = pd.crosstab(df['OverTime'], df['Attrition'], margins=True)
print(overtime_summary)
```

```python
dept_summary['Attrition_Rate_%'] = (dept_summary['Yes'] / dept_summary['All']) * 100
dept_summary
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='OverTime', hue='Attrition', palette='Set2')
plt.title('Attrition Impact by Overtime Status', fontsize=14, fontweight='bold')
plt.xlabel('Works overtime?', fontsize=12)
plt.ylabel('Number of Employees', fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df, 
    x='Attrition', 
    y='Monthly_Income_in$',
    palette='YlGnBu'
)
plt.title('Monthly Income Distribution by Attrition Status', fontsize=14, fontweight='bold')
plt.xlabel('Attrition')
plt.ylabel('Monthly_Income_in$')
plt.show()
```

```python
attrition_counts = df['Attrition'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(attrition_counts, labels=attrition_counts.index, autopct='%1.1f%%', colors=['#4C72B0', '#DD8452'], startangle=90)
plt.title("Overall Employee Attrition Rate")
plt.show()
```

```python
ot_rates = df.groupby('OverTime')['Attrition'].apply(lambda x: (x == 'Yes').mean() * 100)
plt.figure(figsize=(8, 5))
plt.bar(ot_rates.index, ot_rates.values, color=['#4C72B0', '#DD8452'], width=0.5)
plt.title("Attrition Rate by Overtime ")
plt.xlabel("OverTime")
plt.ylabel('Attrition Rate (%)')

plt.show()                                                     
```

```python
role_summary = pd.crosstab(df['JobRole'], df['Attrition'], normalize='index') * 100
role_rates = role_summary['Yes'].sort_values(ascending=True)
plt.figure(figsize=(10, 6))
plt.barh(role_rates.index, role_rates.values, color='#DD8452')
plt.title('Attrition By Job Role')
plt.xlabel('Attrition Rate%')
plt.ylabel('Job Role')
plt.tight_layout()
plt.show()
```

```python
income_avg = df.groupby('Attrition')['Monthly_Income_in$'].mean()
plt.figure(figsize=(6, 5))
plt.bar(income_avg.index, income_avg.values, color=['#4C72B0', '#DD8452'], width=0.4)
plt.title("Attrition by Monthly Income")
plt.xlabel("Attrition")
plt.ylabel("Monthly Income in $")
plt.show()
```

```python
tenure_attrition = pd.crosstab(df['YearsAtCompany'], df['Attrition'])['Yes']
plt.figure(figsize=(10, 5))
plt.plot(tenure_attrition.index, tenure_attrition.values, marker='o', color='#DD8452', linewidth=2)
plt.title("Attrition count by years at a company")
plt.xlabel('Years at a company')
plt.ylabel(' Number of Attrition')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
